```python
#4-WAP to compute frequency distribution
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 200)

plt.hist(x)
plt.show()
```

```python
#5-WAP to compute average and variance
import statistics as s
velocity=[50,70,80,90,57,68,91,77,61,59]
x=np.average(velocity)
print("Average is:",x)
y=s.variance(velocity)
print("Variance is:",y)
z=np.std(velocity)
print("Standard Deviation:",z)
```

```python
#6-WAP to draw normal curves

from scipy.stats import norm

x = np.linspace(-4, 4, 1000)
y = np.exp(-x**2 / 2) / np.sqrt(2 * np.pi)
plt.plot(x, norm.pdf(x), 'b-', label='Normal Distribution')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```
