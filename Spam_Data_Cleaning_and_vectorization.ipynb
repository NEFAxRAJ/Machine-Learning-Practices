```python
## Data Pre-processing
    #1. Lower Case
    #2. Tokenization
    #3. Removing special characters
    #4. Removing stop words and punctuation
    #5. Stemming
```

```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
from nltk.corpus import stopwords
import string
from nltk.stem.porter import PorterStemmer as ps
```

```python
def text_preprocess(text):

  text=text.lower()
  text= nltk.word_tokenize(text)
  y=[]
  for i in text:
    if(i not in stopwords.words('english') and i not in string.punctuation and i.isalnum()):
      y.append(ps().stem(i))
  return " ".join(y)


```

```python
text="Hi how are you not dancing today?"
txt=text_preprocess(text)
print(txt)
```

```python
#Creating a new column of transformed text
data['transformed_text']=data['text'].apply(text_preprocess)
data.head(3)
```

```python
#Vectorizing the data
from sklearn.feature_extraction.text import TfidfVectorizer
x=TfidfVectorizer().fit_transform(data['transformed_text']).toarray()
y=data['type'].values
```

```python
#Splitting the dataset into Train And Test Dataset
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.8,random_state=2)
```
