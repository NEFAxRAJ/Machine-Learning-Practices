SVM Decision Function- Hyperplane

$$f(x) = w^\top x + b
$$

Hinge Loss

$$
\mathcal{L}(x_i, y_i)
= \max\left(0,\; 1 - y_i (w^\top x_i + b)\right)
$$

if: y(wx+b)>=1,

    Loss is zero, meaning point is correctly classified.
else if:y(wx+b)<1,

the loss is positive, indicating a margin violation or misclassification

Soft-margin Cost function

$$ J(w,b) = \;
\frac{1}{2}\|w\|^2
+ C \sum_{i=1}^{n}
\max\left(0,\; 1 - y_i (w^\top x_i + b)\right)
$$

```python
def hinge_loss(x, y, w, b):
    return max(0, 1 - y * (w * x + b))
```

```python
def cost_function(X, y, w, b, C):
    loss = 0
    for i in range(len(X)):
        loss += hinge_loss(X[i][0], y[i], w, b)
    return 0.5 * w * w + C * loss
```

```python
def train_svm(X, y, lr=0.01, C=1, epochs=1000):
    w = 0.0
    b = 0.0

    for _ in range(epochs):
        for i in range(len(X)):
            x_i = X[i][0]
            y_i = y[i]

            if y_i * (w * x_i + b) >= 1:
                w = w - lr * w
            else:
                w = w - lr * (w - C * y_i * x_i)
                b = b + lr * C * y_i
    return w, b
```

```python
documents = [
    "good movie nice",              # class 0
    "excellent wonderful film",     # class 0
    "bad terrible boring movie",    # class 1
    "worst disgusting bad movie",   # class 1
    "average okay not great",       # class 1
    "fine normal decent"            # class 0
]

labels = [0, 0, 1, 1, 1, 0]
```

```python
y = np.array(labels)
y = np.where(y==0, -1, 1)
N = len(y)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer= TfidfVectorizer(ngram_range=(1,2), binary=True, max_features=5)
x=vectorizer.fit_transform(documents).toarray()
```

```python
learning_rate = 0.01
lambda_param = 0.01
epochs = 1000

w = np.zeros(x.shape[1])
b = 0
```

```python
for _ in range(epochs):
    for i in range(len(x)):
        condition = y[i] * (np.dot(x[i], w) + b)

        if condition >= 1:
            w -= learning_rate * (2 * lambda_param * w)
        else:
            w -= learning_rate * (2 * lambda_param * w - y[i] * x[i])
            b -= learning_rate * (-y[i])
```

```python
def predict(text):
    x = vectorizer.transform([text]).toarray()[0]
    score = np.dot(x, w) + b
    return 0 if score < 0 else 1
```

```python
print(predict("good nice film"))
print(predict("bad boring movie"))
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
nltk.download('punkt_tab')
import string
from nltk.stem.porter import PorterStemmer as ps
from bs4 import BeautifulSoup
```

```python
def text_process(text):
  soup= BeautifulSoup(text,"html.parser")
  text=soup.get_text()
  text=text.lower()
  final_txt=""
  for i in nltk.word_tokenize(text):
    if(i not in stopwords.words('english')and i not in string.punctuation and i.isalnum()):
      final_txt+=" "+ps().stem(i)
  return final_txt;

```

```python
df['transformed_review']=df['review'].apply(text_process)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer= TfidfVectorizer(ngram_range=(1,2), binary=True, max_features=50)
x=vectorizer.fit_transform(df['transformed_review']).toarray()
y = df['sentiment'].to_numpy()
```

```python

y = np.where(y==0, -1, 1)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2,random_state=2)
```

```python
def train_svm_model(epochs,x,y, lambda_param):
  w=np.zeros(x.shape[1])
  b = 0
  for _ in range(epochs):
    for i in range(len(x)):
        condition = y[i] * (np.dot(x[i], w) + b)

        if condition >= 1:
            w -= learning_rate * (2 * lambda_param * w)
        else:
            w -= learning_rate * (2 * lambda_param * w - y[i] * x[i])
            b -= learning_rate * (-y[i])
  return w,b
```

```python
learning_rate = 0.01
lambda_param = 0.01
epochs = 1000


w, b =train_svm_model(epochs,x_train,y_train, lambda_param)
```

```python
def predicting_svm(x,w,b):
  y_pred=[]
  for i in x:
    score = np.dot(i, w) + b
    if score < 0:
      y_pred.append(-1)
    else:
      y_pred.append(1)
  return y_pred
```

```python
y_pred=predicting_svm(x_test,w,b)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
