### **B.8. Implement KNN using Python-based ML tools .**

K-Nearest Neighbours (K-NN) is a supervised, non-parametric machine learning algorithm used for classification and regression.


To predict a new data point, K-NN:

Finds the K closest data points (neighbors) in the training data

Uses their labels to make a prediction

Classification → majority vote

Regression → average of values

To calculate the distance we use Euclidean distance.

$$
d(x,y) = \sqrt{\sum_{i=1}^{n} (x_i - y_i)^2}
$$


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Here, we are building 2-Dimensional K-NN model

```python
data= {
    "x_axis":[1,2,3,6,7,8],
"y_axis":[2,3,1,6,7,5],
       "label":['A','A','A','B','B','B']
}

df=pd.DataFrame(data)
df.head()
```

*The euclidean dstance to calculate thedistance between points*

```python
def euclidean_distance(a,b):
    return np.sqrt(np.sum(a-b)**2)
```

```python
def knn_predict(df, query_point, k):
    distances = []

    for i in range(len(df)):
        point = df.iloc[i][["x_axis", "y_axis"]].values.astype(float)
        label = df.iloc[i]["label"]

        dist = euclidean_distance(point, query_point)
        distances.append((dist, label))

    # Sort by distance
    distances.sort(key=lambda x: x[0])

    # Select k nearest neighbors
    k_nearest = distances[:k]

    # Majority vote
    labels = [label for _, label in k_nearest]
    prediction = max(set(labels), key=labels.count)

    return prediction, k_nearest
```

```python
query = np.array([5,9])
k = 3

prediction, neighbors = knn_predict(df, query, 3)

print("Predicted class:", prediction)
print("Nearest neighbors:", neighbors)
```

### We are making a multi-dimensional K-NN model which classify sentiment of reviews

```python
documents = [
    "good movie nice",              # class 0
    "excellent wonderful film",     # class 0
    "bad terrible boring movie",    # class 1
    "worst disgusting bad movie",   # class 1
    "average okay not great",       # class 1
    "fine normal decent"            # class 0
]

labels = [0, 0, 1, 1, 1, 0]
```

```python
x=documents
y=labels
```

```python
def build_vocab(docs):
    vocab = set()
    for doc in docs:
        for word in doc.lower().split():
            vocab.add(word)
    return sorted(vocab)
```

```python
# This is the set of features/words set for this model
vocab = build_vocab(documents)
```

```python
#
def bow_vector(doc, vocab):
    vector = np.zeros(len(vocab))
    words = doc.lower().split()

    for i, word in enumerate(vocab):
        vector[i] = words.count(word)

    return vector
```

```python
x = np.array([bow_vector(doc, vocab) for doc in documents])
y = np.array(labels)
```

```python
x
```

```python
def euclidean_distance(a,b):
  return np.sqrt(np.sum((a - b) ** 2))
```

```python
def knn_predict2(X, y, query_vector, k):
    distances = []

    for i in range(len(X)):
        # IMPORTANT: calling the function, not storing it
        dist = euclidean_distance(X[i], query_vector)
        distances.append((dist, y[i]))

    # Now dist is numeric → sorting works
    distances.sort(key=lambda x: x[0])

    k_nearest = distances[:k]
    labels = [label for _, label in k_nearest]

    return max(set(labels), key=labels.count)
```

```python
text = "it is nice film"
query_vec = bow_vector(text, vocab)

prediction = knn_predict2(x, y, query_vec, k=3)
if(prediction==0):
  print("Predicted Sentiment: Positive")
else:
  print("Predicted Sentiment: Negative")
```
