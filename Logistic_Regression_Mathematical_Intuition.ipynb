**Logistic Regression using gradient descent- Mathematical Intuition**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
  return (1/(1+np.exp(-z)))
```

```python
def hypothesis(Q0,Q1,x):
  h=Q0+Q1*x
  return h
```

```python
def costfunction(Q0,Q1,x,y):
  m=len(x)
  cost=0
  for i in range(0,m):
    h=sigmoid(hypothesis(Q0,Q1,x[i]))
    cost= cost + (y[i]*np.log(h)+(1-y[i])*np.log(1-h))
  return ((-1/m)*cost)
```

```python
def gradient_function(Q0,Q1,x,y):
  m=len(x)
  q0,q1=0,0
  for i in range(m):
    h=sigmoid(hypothesis(Q0,Q1,x[i]))
    q0+=(h-y[i])
    q1+=(h-y[i])*x[i]
  Q0=q0/m
  Q1=q1/m
  return (Q0,Q1)
```

```python
def gradient_descent(Q0,Q1,learning_rate,x,y):
  for i in range(2000):
    q0,q1=gradient_function(Q0,Q1,x,y)
    Q0=Q0-learning_rate*q0
    Q1=Q1-learning_rate*q1
    if(i%500==0):
      print("Cost at ", i," th iteration is ",costfunction(Q0,Q1,x,y))
      print(Q0,Q1)
  return Q0,Q1;
```

```python
Q0,Q1=0,0
x=np.array([1,2,3,4,5,6,7,8])
y=np.array([0,0,0,0,0,1,1,1,])
learning_rate=0.1
Q0,Q1=gradient_descent(Q0,Q1,learning_rate,x,y)
```

```python
plt.scatter(x, y, label='Training data')

m = len(x)
predicted_y = []

for i in range(m):
    h = sigmoid(hypothesis(Q0,Q1,x[i]))
    predicted_y.append(h)


plt.plot(x, predicted_y, color='red', label='Logistic Curve')

plt.xlabel("x")
plt.ylabel("Predicted Value")
plt.title("Logistic Regression Curve")
plt.legend()
plt.show()
```
